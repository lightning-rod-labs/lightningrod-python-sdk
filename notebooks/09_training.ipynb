```python
from dotenv import load_dotenv
from lightningrod import LightningRod
from lightningrod.training import TrainingConfig
from lightningrod.utils import config

load_dotenv()
api_key = config.get_config_value("LIGHTNINGROD_API_KEY")

lr = LightningRod(api_key=api_key)

config = TrainingConfig(
    input_dataset_id="cfb63189-a655-4d24-96f6-45a6ed181b53",
    name="Startup forecaster",
    training_steps=3,
)
```

```python
cost = lr.training.estimate_cost(config)
print(f"Estimated cost: ${cost.total_cost_dollars:.2f}")
print(f"Effective steps: {cost.effective_steps}, Tokens: prefill={cost.prefill_tokens}, sample={cost.sample_tokens}, train={cost.train_tokens}")
```

```python
job = lr.training.run(config)
print(f"Training job finished: {job.id}: {job.status}")
```

```python
%pip install openai

from openai import OpenAI

client = OpenAI(
    base_url=f"{lr.base_url}/openai",
    api_key=api_key,
)
```

```python
response = client.chat.completions.create(
    model="snapshot:9b54852c-ffd7-4b59-88ed-784b2d47a891",
    messages=[
        {"role": "system", "content": "You are a startup forecaster."},
        {"role": "user", "content": "Will Anthropic's annual recurring revenue exceed $2B by the end of 2026?"},
    ],
)
print(response.choices[0].message.content)
```
