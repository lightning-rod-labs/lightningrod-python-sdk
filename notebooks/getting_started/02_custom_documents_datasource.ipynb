# Custom Document Dataset Generation

This notebook demonstrates how to generate forecasting questions from your own documents and files. You can use PDFs, text, CSV or other types of files as a data source for question generation.

## Set up the client

Sign up at [dashboard.lightningrod.ai](https://dashboard.lightningrod.ai/sign-up?redirect=/api) to get your API key.

- **Google Colab**: Go to the Secrets section (key icon in left sidebar) and add a secret named `LIGHTNINGROD_API_KEY`
- **Local Jupyter**: Set the `LIGHTNINGROD_API_KEY` environment variable, or you'll be prompted to enter it

```python
from dotenv import load_dotenv
from lightningrod import LightningRod
from lightningrod.utils import config

load_dotenv()
api_key = config.get_config_value("LIGHTNINGROD_API_KEY")

lr = LightningRod(api_key=api_key)
```

## Create Input Dataset

Now we'll create Sample objects from our input files and upload them to create an input dataset that can be used for question generation.

> Tip: if you prefer more control, you can also generate these samples manually by chunking the input file(s) according to your needs.

```python
from lightningrod import preprocessing

# Required by lightningrod.preprocessing
%pip install langchain-text-splitters
!git clone https://github.com/lightning-rod-labs/lightningrod-python-sdk

from IPython.display import clear_output
clear_output()

samples = preprocessing.files_to_samples("./lightningrod-python-sdk/notebooks/getting_started/data/reddit_content_policy.pdf",)
print(f"Created {len(samples)} samples from chunks")

input_dataset = lr.datasets.create_from_samples(samples, batch_size=1000)
print(f"Created input dataset: {input_dataset.id}")
print(f"Total samples: {input_dataset.num_rows}")
```

## Configure Question Generator

Generate questions based on the content of your documents.

```python
from lightningrod import BinaryAnswerType, QuestionAndLabelGenerator

answer_type = BinaryAnswerType()

qa_config = QuestionAndLabelGenerator(
    answer_type=answer_type,
    questions_per_seed=2,
    instructions=(
        "Generate binary forecasting questions based on Reddit Content Policy content. "
        "Focus on specific future outcomes related to policy enforcement, rule changes, community moderation actions, or transparency reporting. "
        "Questions must reference concrete rules, enforcement mechanisms, or policy provisions mentioned in the document. "
        "Questions must be complete and stand-alone - they should not reference external information like rule numbers, document names, or 'as stated in X' without explaining what X is. "
    ),
    examples=[
        "Will Reddit remove content that reveals confidential information about a user without their consent?",
        "Will Reddit ban accounts that repeatedly violate the platform's anti-harassment policies?",
        "Will Reddit permit the posting of content that exposes personal addresses even if the user objects?",
        "Will Reddit allow users to share explicit imagery without any content warnings or labeling?",
    ],
    bad_examples=[
        "Will Reddit ban communities that incite violence or promote hate based on identity or vulnerability as stated in Rule 1? (bad: references 'Rule 1' without explaining what it is - question won't be understandable without the original document)",
        "Will Reddit add NSFW tags or quarantine restrictions to communities that violate Rule 7's content labeling requirements? (bad: references 'Rule 7' without explaining what it is - question requires external knowledge)",
        "Will Reddit remove content that violates Rule 3 (respecting privacy) by revealing someone's personal information? (bad: references 'Rule 3' without explaining what it is - question won't be understandable without the original document)",
        "Will Reddit publish an updated Transparency Report as referenced in the document? (bad: references 'the document' - question assumes reader has access to the source document)",
    ],
)
```

## Run the Pipeline with Input Dataset

Pass our input dataset to the transform pipeline. The samples from the input dataset will be used as seeds for question/answer generation.

```python
chunk_dataset = lr.transforms.run(
    qa_config, 
    input_dataset=input_dataset,
    max_seeds=5,  # keep low when testing; increase to ~5000 for a real run
    name="Custom documents"
)
```

> Note: This can take a few minutes to complete processing.
