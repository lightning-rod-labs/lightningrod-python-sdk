# Create a Custom FileSet

A **FileSet** is a collection of documents with optional metadata that you can use as a data source for question generation. This notebook walks through creating a FileSet, defining a metadata schema, and uploading documents.

We'll create synthetic quarterly investor reports for two companies — APEX Technologies and Vanguard Industries (VGI) — spanning Q1 2024 through Q1 2025. Notebooks 2 and 3 in this series use this FileSet for question generation and advanced features.

## Create FileSet with Metadata Schema

We define a metadata schema with two fields:
- **ticker** — the company ticker symbol (e.g. `APEX`, `VGI`)
- **quarter** — the fiscal quarter (e.g. `Q1 2024`)

These metadata fields can later be used to filter which documents are included in question generation.

```python
from lightningrod import (
    FileSetMetadataSchemaInput,
    MetadataFieldDefinitionInput,
    MetadataFieldType,
)

schema = FileSetMetadataSchemaInput(fields=[
    MetadataFieldDefinitionInput(
        name="ticker", field_type=MetadataFieldType.STRING, required=True,
        description="Company ticker symbol"
    ),
    MetadataFieldDefinitionInput(
        name="quarter", field_type=MetadataFieldType.STRING, required=True,
        description="Fiscal quarter (e.g. Q1 2024)"
    ),
])

fileset = lr.filesets.create(
    name="Quarterly Reports - APEX & VGI",
    description="Synthetic quarterly investor reports for APEX Technologies and Vanguard Industries (Q1 2024 - Q1 2025).",
    metadata_schema=schema,
)
print(f"Created FileSet: {fileset.id}")
print(f"  Name: {fileset.name}")
```

## Upload Documents

The SDK provides two high-level methods for uploading files:
- **`upload_files()`** — upload a list of file paths with optional metadata
- **`upload_directory()`** — upload all files from a directory matching a pattern

For this example, we'll write the documents to temporary files and upload them.

```python
import tempfile
from pathlib import Path

# Write documents to temporary files
temp_dir = Path(tempfile.mkdtemp())
file_paths = []
metadata = {}

for doc in documents:
    # Write content to temp file
    file_path = temp_dir / doc["filename"]
    file_path.write_text(doc["content"])
    file_paths.append(file_path)
    
    # Build metadata for this file
    metadata[doc["filename"]] = {
        **doc["metadata"],
        "file_date": doc["file_date"],  # SDK handles datetime conversion
    }

print(f"Created {len(file_paths)} temp files in {temp_dir}")

# Upload all files with metadata in one call
result = lr.filesets.upload_files(
    fileset.id,
    file_paths,
    metadata=metadata,
)

print(f"\n✓ Uploaded {result.succeeded} files")
if result.failed > 0:
    print(f"✗ Failed: {result.failed}")
    for error in result.errors[:5]:
        print(f"  - {error}")

# Clean up temp files
import shutil
shutil.rmtree(temp_dir)
```
